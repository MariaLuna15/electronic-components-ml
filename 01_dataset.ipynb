```python
from PIL import Image
import os

def limpiar_dataset(directorio, min_size=50):
    eliminadas = 0
    for clase in os.listdir(directorio):
        ruta = os.path.join(directorio, clase)
        if not os.path.isdir(ruta):
            continue
        for fname in os.listdir(ruta):
            path = os.path.join(ruta, fname)
            try:
                img = Image.open(path)
                w, h = img.size
                if w < min_size or h < min_size:
                    os.remove(path)
                    eliminadas += 1
            except Exception:
                os.remove(path)
                eliminadas += 1
    print(f"✓ Limpieza completa — {eliminadas} imágenes eliminadas")

limpiar_dataset('raw/')
```

```python
import shutil, random
from pathlib import Path

def dividir_dataset(src, dst, ratios=(0.70, 0.15, 0.15), seed=42):
    random.seed(seed)
    splits = ['train', 'val', 'test']

    for clase in Path(src).iterdir():
        if not clase.is_dir():
            continue
        imagenes = list(clase.glob('*.jpg')) + list(clase.glob('*.png')) + list(clase.glob('*.jpeg'))
        random.shuffle(imagenes)

        n = len(imagenes)
        c1 = int(n * ratios[0])
        c2 = int(n * (ratios[0] + ratios[1]))
        grupos = [imagenes[:c1], imagenes[c1:c2], imagenes[c2:]]

        for split, grupo in zip(splits, grupos):
            destino = Path(dst) / split / clase.name
            destino.mkdir(parents=True, exist_ok=True)
            for img in grupo:
                shutil.copy(img, destino / img.name)

        print(f"{clase.name:15} → train: {len(grupos[0])} | val: {len(grupos[1])} | test: {len(grupos[2])}")

dividir_dataset('raw/', 'dataset/')
print("\n✓ Dataset listo")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os, random

clases = ['resistencia', 'capacitor', 'led', 'transistor', 'diodo']
fig, axes = plt.subplots(5, 4, figsize=(12, 14))

for i, clase in enumerate(clases):
    ruta = f'dataset/train/{clase}'
    imagenes = os.listdir(ruta)
    muestra = random.sample(imagenes, min(4, len(imagenes)))
    for j, fname in enumerate(muestra):
        img = Image.open(os.path.join(ruta, fname)).convert('RGB')
        axes[i, j].imshow(img)
        axes[i, j].axis('off')
        axes[i, j].set_title(clase, fontsize=9)

plt.tight_layout()
plt.savefig('muestra_dataset.png', dpi=100)
plt.show()
print("✓ Vista previa guardada")
```

```python
from icrawler.builtin import BingImageCrawler
import os, shutil

componentes_extra = {
    'diodo':       ['diode real photo close up', 'silicon diode 1N4007', 'rectifier diode black component'],
    'capacitor':   ['electrolytic capacitor blue', 'tantalum capacitor component', 'capacitor PCB real'],
    'led':         ['LED 5mm red green blue', 'through hole LED component', 'LED indicator light component'],
    'resistencia': ['carbon film resistor bands', 'through hole resistor PCB', 'resistor component macro'],
    'transistor':  ['BC547 transistor black', 'TO-92 transistor component', 'BJT transistor real photo']
}

for clase, keywords in componentes_extra.items():
    carpeta_temp = f'raw_extra/{clase}'
    os.makedirs(carpeta_temp, exist_ok=True)
    print(f"\nDescargando: {clase}...")
    for keyword in keywords:
        crawler = BingImageCrawler(storage={'root_dir': carpeta_temp})
        crawler.crawl(keyword=keyword, max_num=80)
    
    # Mover a raw con nombres nuevos
    archivos = os.listdir(carpeta_temp)
    existentes = len(os.listdir(f'raw/{clase}'))
    nuevos = 0
    for i, fname in enumerate(archivos):
        src = os.path.join(carpeta_temp, fname)
        dst = os.path.join(f'raw/{clase}', f'extra_{i+existentes:04d}.jpg')
        shutil.copy(src, dst)
        nuevos += 1
    print(f"  ✓ {nuevos} imágenes nuevas agregadas a {clase}")

print("\n✓ Descarga extra completa")
```
